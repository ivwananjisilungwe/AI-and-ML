## 🧠 Intuition: Two Operations That Define Linear Algebra

There are only **two** fundamental operations in all of linear algebra. Everything else is built from these.

### 1. Scaling (multiply a vector by a number)

$$2 \cdot \begin{bmatrix} 1 \\ 2 \end{bmatrix} = \begin{bmatrix} 2 \\ 4 \end{bmatrix}$$

- If $c > 1$: the arrow gets **longer** (same direction)  
- If $0 < c < 1$: the arrow gets **shorter** (same direction)  
- If $c < 0$: the arrow **flips direction**  
- If $c = 0$: you get the **zero vector** $\begin{bmatrix}0 \\ 0\end{bmatrix}$

### 2. Addition (add two vectors)

$$\begin{bmatrix} 1 \\ 2 \end{bmatrix} + \begin{bmatrix} 3 \\ 1 \end{bmatrix} = \begin{bmatrix} 4 \\ 3 \end{bmatrix}$$

Geometrically: place the tail of the second arrow at the **tip** of the first. The result is the vector from the origin to the final point. This is called **tip-to-tail** addition.

### Putting them together: Linear Combination

A **linear combination** is: **scale then add**.

$$c_1 \vec{v}_1 + c_2 \vec{v}_2 + \dots + c_n \vec{v}_n$$

> **This is THE most important idea in linear algebra.** Almost everything else (matrix multiplication, dot products, transformations) is a variation of this.

## 1. Vector Scaling in NumPy

```python
import numpy as np

v = np.array([2, 1])

print("v    =", v)
print("2*v  =", 2 * v)    # Scale by 2 → longer, same direction
print("0.5*v =", 0.5 * v)  # Scale by 0.5 → shorter, same direction
print("-1*v =", -1 * v)    # Scale by -1 → opposite direction, same length
print("-2*v =", -2 * v)    # Scale by -2 → opposite direction, longer
print()

# Scaling is element-wise — each element gets multiplied
print(f"  v    = [{v[0]}, {v[1]}]")
print(f"  3*v  = [{3*v[0]}, {3*v[1]}]")
```

### 🔍 Discover the Scaling Pattern

**Before running the code, think about this:**

If $\vec{v} = \begin{bmatrix}3 \\ 4\end{bmatrix}$:
- What is $2\vec{v}$?  
- What is $0.5\vec{v}$?  
- What is $-3\vec{v}$?

All three lie on the **same line** through the origin.  
The only difference is **how far from the origin they reach** — their length changes.

```python
v = np.array([3, 4])

print("v     =", v)
print("2*v   =", 2*v)
print("0.5*v =", 0.5*v)
print("-3*v  =", -3*v)
print()

print("Notice: they all point along the same line!")
```

## 2. Vector Addition in NumPy

Adding two vectors: add **corresponding elements**.

```python
a = np.array([1, 2])
b = np.array([3, 1])

print("a =", a)
print("b =", b)
print("a + b =", a + b)    # [1+3, 2+1] = [4, 3]
print()

# Works for any dimension (must be the SAME dimension!)
x = np.array([10, 20, 30])
y = np.array([1, 2, 3])
print("x + y =", x + y)    # [11, 22, 33]
```

**Question:** What happens if you try to add vectors of different dimensions?

Try it: `np.array([1,2]) + np.array([1,2,3])`  

> *(Answer: NumPy will raise an error — you can only add vectors of the same shape)*

```python
try:
    np.array([1, 2]) + np.array([1, 2, 3])
except ValueError as e:
    print("Error:", e)
    print("You can only add vectors of the same dimension!")
```

## 3. Linear Combinations

Scale **then** add: $c_1\vec{a} + c_2\vec{b}$

```python
a = np.array([1, 2])
b = np.array([3, 1])

# Linear combination: 2*a + 3*b
c1, c2 = 2, 3
result = c1 * a + c2 * b

print(f"{c1}*a + {c2}*b = {c1}*{a} + {c2}*{b}")
print(f"              = {c1*a} + {c2*b}")
print(f"              = {result}")
print()

# Try different coefficients
print("0.5*a + 2*b  =", 0.5 * a + 2 * b)
print("-1*a + 1*b  =", -1 * a + 1 * b)
print("3*a + 0*b    =", 3 * a + 0 * b)  # b contributes nothing
```

### Properties of Linear Combinations

1. **Distributive:** $c(\vec{a} + \vec{b}) = c\vec{a} + c\vec{b}$  
2. **Zero vector:** $\vec{v} + \vec{0} = \vec{v}$  
3. **Commutative:** $\vec{a} + \vec{b} = \vec{b} + \vec{a}$  
4. **Associative:** $(\vec{a} + \vec{b}) + \vec{c} = \vec{a} + (\vec{b} + \vec{c})$

```python
a = np.array([1, 2]); b = np.array([3, 1]); c = 2.5

# Property 1: Distributive
lhs = c * (a + b)
rhs = c * a + c * b
print(f"c*(a+b)   = {lhs}")
print(f"c*a + c*b = {rhs}")
print(f"Distributive holds? {np.allclose(lhs, rhs)} ✅")
```

## ✏️ Exercise 2 — ML Context

A linear model predicts house price:

$$\text{price} = w_1 \cdot \text{size} + w_2 \cdot \text{bedrooms} + w_3 \cdot \text{age} + b$$

Features: $\vec{x} = \begin{bmatrix}120 & 3 & 10\end{bmatrix}$  
Weights: $\vec{w} = \begin{bmatrix}2.5 & 10 & -1\end{bmatrix}$  
Bias: $b = 50$

Compute the predicted price using a **linear combination** of features and weights.

```python
x = np.array([120, 3, 10])
w = np.array([2.5, 10, -1])
b = 50

# Linear combination: w1*x1 + w2*x2 + w3*x3 + b
price = w[0]*x[0] + w[1]*x[1] + w[2]*x[2] + b
print(f"Predicted price: ${price:.2f} thousand")

# Or use numpy's dot product (same thing, cleaner)
price2 = np.dot(w, x) + b
print(f"Using dot product: ${price2:.2f} thousand")
```
