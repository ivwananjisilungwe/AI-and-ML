## 📚 Week 1 — All Concepts Review

### Day 1: What is a vector?
- A container for numbers (data view) or an arrow (geometry view)
- In ML, every data point is a vector
- NumPy: `np.array()`, `.shape`, indexing, slicing

### Day 2: Scaling & Addition (Linear Combinations)
- Scaling: $c \cdot \vec{v}$ — multiply each element
- Addition: $\vec{a} + \vec{b}$ — add element-wise
- Linear combination: $c_1\vec{v}_1 + c_2\vec{v}_2$ — scale + add
- ML: $\text{prediction} = \vec{w} \cdot \vec{x} + b$

### Day 3: The Span of Vectors
- Span = set of ALL points reachable via linear combinations
- One vector → line. Two non-colinear → plane. Colinear → line.
- ML: feature space, intrinsic dimensionality

### Day 4: Parametric Lines
- $\text{Line}(t) = \vec{p} + t\vec{v}$
- Line through two points: $\vec{a} + t(\vec{b} - \vec{a})$
- ML: decision boundaries, gradient descent, interpolation

---
## 💡 The Big Idea of Week 1

**Every problem in linear algebra comes down to:**

> "Given a set of vectors, can I find coefficients that combine them to reach a target?"

$$c_1\vec{v}_1 + c_2\vec{v}_2 + \dots + c_n\vec{v}_n = \vec{b}$$

- If $\vec{b}$ is **in the span** → exact solution  
- If $\vec{b}$ is **not in the span** → find the closest solution (least squares — Week 12)

---
## 🔄 From Linear Combinations to Matrix Equations

This:
$$c_1\begin{bmatrix}1 \\ 2\end{bmatrix} + c_2\begin{bmatrix}3 \\ 1\end{bmatrix} = \begin{bmatrix}7 \\ 8\end{bmatrix}$$

Becomes this:
$$\underbrace{\begin{bmatrix}1 & 3 \\ 2 & 1\end{bmatrix}}_{A} \underbrace{\begin{bmatrix}c_1 \\ c_2\end{bmatrix}}_{\vec{c}} = \underbrace{\begin{bmatrix}7 \\ 8\end{bmatrix}}_{\vec{b}}$$

Or simply: $A \cdot \vec{c} = \vec{b}$  
Solve with: `c = np.linalg.solve(A, b)`

## 🎯 Mini-Exercise: "Reach the Target"

### Scenario

Your drone can move in two directions:

$$\vec{v}_1 = \begin{bmatrix}1 \\ 2\end{bmatrix}, \quad \vec{v}_2 = \begin{bmatrix}3 \\ 1\end{bmatrix}$$

Your target:

$$\vec{b} = \begin{bmatrix}7 \\ 8\end{bmatrix}$$

Find $c_1, c_2$ such that $c_1\vec{v}_1 + c_2\vec{v}_2 = \vec{b}$.

**Try solving by hand first (system of equations):**

1. $1c_1 + 3c_2 = 7$
2. $2c_1 + 1c_2 = 8$

> *Solve it, then run the code to check.*

```python
import numpy as np

v1 = np.array([1, 2]); v2 = np.array([3, 1]); target = np.array([7, 8])

# Manual (from substitution)
c1, c2 = 3.4, 1.2
result = c1*v1 + c2*v2
print(f"Manual: c1={c1}, c2={c2}")
print(f"Result: {result}")
print(f"Target: {target}")
print(f"Match? {np.allclose(result, target)} ✅")
```

### Now with `np.linalg.solve`

```python
A = np.column_stack((v1, v2))
print("Matrix A (columns = v1, v2):")
print(A)
print()

c = np.linalg.solve(A, target)
print(f"Solution: c1={c[0]:.2f}, c2={c[1]:.2f}")

reached = A @ c
print(f"A @ c = {reached}")
print(f"✓ Exact match! {np.allclose(reached, target)}")
```

### Visualizing the Linear Combination

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,8))
plt.axhline(0, color='gray', linewidth=0.5); plt.axvline(0, color='gray', linewidth=0.5)

plt.arrow(0,0,v1[0],v1[1], head_width=0.2,head_length=0.3, color='red', width=0.05, label='v1')
plt.arrow(0,0,v2[0],v2[1], head_width=0.2,head_length=0.3, color='blue', width=0.05, label='v2')

step1 = c[0]*v1
plt.arrow(0,0,step1[0],step1[1], head_width=0.15,head_length=0.2, color='orange', width=0.03, linestyle='--', label=f'{c[0]:.1f}*v1')
plt.arrow(step1[0],step1[1],v2[0]*c[1],v2[1]*c[1], head_width=0.15,head_length=0.2, color='purple', width=0.03, linestyle='--', label=f'{c[1]:.1f}*v2')

plt.scatter(target[0],target[1], color='green', s=200, zorder=5, label='Target')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-1,10); plt.ylim(-1,10)
plt.title(f'{c[0]:.1f}·v1 + {c[1]:.1f}·v2 = Target')
plt.legend(); plt.show()
```

## 🏆 Week 1 Validation Challenge

### Part 1: Solve a System

$$\vec{a} = \begin{bmatrix}2 \\ 1\end{bmatrix}, \;
\vec{b} = \begin{bmatrix}-1 \\ 3\end{bmatrix}, \;
\text{target} = \begin{bmatrix}4 \\ 11\end{bmatrix}$$

1. Solve using `np.linalg.solve`  
2. Verify by computing the linear combination  
3. Plot the result

```python
a = np.array([2,1]); b = np.array([-1,3]); target = np.array([4,11])
A = np.column_stack((a,b))
c = np.linalg.solve(A, target)

print(f"c1 = {c[0]:.2f}, c2 = {c[1]:.2f}")
print(f"c1*a + c2*b = {c[0]*a + c[1]*b}")
print(f"Target      = {target}")
print(f"✅ Correct? {np.allclose(c[0]*a + c[1]*b, target)}")
```

### Part 2: When There's No Solution

$$\vec{a} = \begin{bmatrix}1 \\ 1\end{bmatrix}, \;
\vec{b} = \begin{bmatrix}2 \\ 2\end{bmatrix} \quad \text{(colinear!)}, \;
\text{target} = \begin{bmatrix}5 \\ 3\end{bmatrix}$$

**❓ Before running: why will this fail?**

```python
A = np.column_stack(([1,1], [2,2]))
print("Det(A) =", np.linalg.det(A), "← matrix is SINGULAR")
print("(5,3) is NOT on the line y=x — no combination of colinear vectors can reach it.")

try:
    np.linalg.solve(A, [5,3])
except np.linalg.LinAlgError as e:
    print(f"❌ Error: {e}")
```
