## 🧠 Intuition: What is a Vector?

Imagine you are building an ML model that predicts **house prices**.

A house has features:
- Size: 120 square meters
- Bedrooms: 3
- Age: 10 years
- Distance from city: 5 km

A computer cannot understand "house". It only understands numbers.  
So we represent the house as a **vector**:

$$\text{House} = \begin{bmatrix} 120 \\ 3 \\ 10 \\ 5 \end{bmatrix}$$

**A vector is just a way of storing multiple pieces of information together.**

In ML, vectors can represent:
- 🏠 A house (features)
- 👤 A person (age, height, weight)
- 🖼️ An image (millions of pixels flattened into one list)
- 📝 A word ("king" → vector of 300 numbers)
- ⭐ A user's preferences (ratings for movies)

### 👁️ Visual Intuition — The Arrow

Think of a vector as an arrow.

$\begin{bmatrix} 3 \\ 2 \end{bmatrix}$ means "Move 3 steps horizontally and 2 steps vertically."

```
      y
      ↑
      |
     2 |       ● (3,2)
      |      /|
      |     / |
      |    /  |
      |   /   |
      |  /    |
      | /     |
     0 └─────┴─────→ x
           3
```

Every arrow (vector) has **two** important properties:
1. **Direction** — where the arrow points
2. **Magnitude (length)** — how long the arrow is

> ⭐ **Key insight:** Both views are correct — the **geometry view** (arrow) and the **data view** (list of numbers). Which one you use depends on the problem.

## ✅ Check Your Understanding

**Question:** If a vector is $\begin{bmatrix}3 & 4\end{bmatrix}$, what do the numbers 3 and 4 represent?

- A) The number 3 and the number 4  
- B) 3 units in x-direction, 4 units in y-direction (movement)  
- C) 3 and 4 are features describing something  
- D) Both B and C are valid interpretations  

> *Think about it before reading on...*

**Answer: D — Both!**  
- **Geometry view:** $[3,4]$ is a movement: 3 units right and 4 units up.  
- **ML/Data view:** $[3,4]$ is a collection of features describing something.

One small correction: a vector $[3,4]$ is **2-dimensional**, so it cannot represent RGB color (that needs 3 components). $[3,4,5]$ could be an RGB color.

## 🔢 Creating Vectors in NumPy

````python
import numpy as np

# Create a vector representing a student's scores [Math, Science, English]
student = np.array([85, 90, 78])

print("Student vector:", student)
print("Type:", type(student))
print()

# Creating more vectors
house = np.array([120, 3, 10, 5])
car = np.array([150, 1400, 2022])

print("House features vector:", house)
print("Car features vector:", car)
````

### 📏 Vector Shape

Every NumPy array has a `.shape` attribute. Let's see what ours look like.

````python
student = np.array([85, 90, 78])

print("Student vector:", student)
print("Shape:", student.shape)
print("Number of elements (size):", student.size)
print()

print("(3,) means: a 1-dimensional array with 3 elements")
print("The comma is Python's way of showing it's a tuple with one value.")
````

**Question:** What does `(3,)` tell us?

- A) The vector has 3 rows and 1 column  
- B) The vector has 3 elements and is one-dimensional  
- C) The vector has 3 dimensions in space  
- D) It is a 3×3 matrix

> *(Answer: B. A 1D array with 3 elements. `(3,1)` would be a 2D column vector.)*

## 🎯 Accessing Elements (Indexing)

Vectors are **0-indexed** — the first element is at position 0.

````python
student = np.array([85, 90, 78])

print("student =", student)
print()
print("student[0] =", student[0], "   ← First element (Math)")
print("student[1] =", student[1], "   ← Second element (Science)")
print("student[2] =", student[2], "   ← Third element (English)")
print()

# Negative indexing (from the end)
print("student[-1] =", student[-1], "  ← Last element")
print("student[-2] =", student[-2], "  ← Second-to-last")
````

---
## ✂️ Slicing Vectors

`start:stop` — the `stop` index is **exclusive** (not included).

````python
data = np.array([10, 20, 30, 40, 50, 60])

print("Full vector:", data)
print()
print("data[0:3]   =", data[0:3], "  ← Elements at index 0, 1, 2")
print("data[:3]    =", data[:3], "   ← Same thing (start defaults to 0)")
print("data[3:]    =", data[3:], "   ← From index 3 to end")
print("data[1:4]   =", data[1:4], "  ← Elements at index 1, 2, 3")
print("data[-3:]   =", data[-3:], "  ← Last 3 elements")
print()

# Step slicing
print("data[::2]   =", data[::2], "   ← Every 2nd element")
print("data[::-1]  =", data[::-1], "  ← Reversed")
````

## 🤖 Why Vectors Matter in Machine Learning

**Almost everything in ML becomes a vector.**

| ML Concept | Vector Representation |
|-----------|----------------------|
| **Input features** | $x = \begin{bmatrix} \text{age} \\ \text{income} \\ \text{credit\_score} \end{bmatrix}$ |
| **Model weights** | $w = \begin{bmatrix} 0.5 \\ -0.1 \\ 0.8 \end{bmatrix}$ |
| **A color pixel** | $\begin{bmatrix} 255 \\ 128 \\ 64 \end{bmatrix}$ (RGB) |
| **Word embedding** | $\begin{bmatrix} 0.12 \\ -0.45 \\ 0.78 \end{bmatrix}$ |
| **Image (224×224)** | A vector of **150,528 numbers** |

A simple neural network computes: $y = w \cdot x + b$  
That's an operation on **vectors** — which we'll learn in the coming days.
