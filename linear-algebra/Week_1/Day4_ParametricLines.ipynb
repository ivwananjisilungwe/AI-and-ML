## 🧠 Intuition: What is a Parametric Line?

A line is just **a point plus all possible movements in a direction**.

### The Formula

$$\text{Line}(t) = \vec{p} + t \cdot \vec{v}$$

| $t$ value | Point | Meaning |
|-----------|-------|---------|
| $t = 0$ | $\vec{p}$ | Starting point |
| $t = 1$ | $\vec{p} + \vec{v}$ | One step in direction v |
| $t = 2$ | $\vec{p} + 2\vec{v}$ | Two steps |
| $t = -1$ | $\vec{p} - \vec{v}$ | One step backward |
| $t = 0.5$ | $\vec{p} + 0.5\vec{v}$ | Halfway there |

> **Key insight:** A parametric line is the **span of $\vec{v}$ shifted by $\vec{p}$**.

## 1. Line Through the Origin

Simplest case: $\vec{p} = \vec{0}$ — this is just the span of $\vec{v}$.

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([1, 2])
t_values = np.linspace(-3, 3, 100)
points = np.array([t * v for t in t_values])

plt.figure(figsize=(6,6))
plt.plot(points[:,0], points[:,1], 'b-', label='Line: t*[1,2]')
plt.axhline(0, color='gray', linewidth=0.5); plt.axvline(0, color='gray', linewidth=0.5)
plt.arrow(0,0,v[0],v[1], head_width=0.15,head_length=0.2, color='red', width=0.03, label='v')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-5,5); plt.ylim(-5,5)
plt.title('Line through origin: $t \\cdot \\vec{v}$')
plt.legend(); plt.show()
```

---
## 2. Line Through Any Point

Now shift the line so it passes through an arbitrary point $\vec{p}$.

```python
p = np.array([1, 2]); v = np.array([2, 1])
t_values = np.linspace(-3, 3, 100)
points = np.array([p + t*v for t in t_values])

plt.figure(figsize=(6,6))
plt.plot(points[:,0], points[:,1], 'b-', label=f'Line: p + t*v')
plt.axhline(0, color='gray', linewidth=0.5); plt.axvline(0, color='gray', linewidth=0.5)
plt.scatter(p[0], p[1], color='green', s=150, label=f'p (t=0)')
plt.arrow(p[0], p[1], v[0], v[1], head_width=0.15, head_length=0.2, color='red', width=0.03, label='v')
plt.scatter(p[0]+v[0], p[1]+v[1], color='orange', s=100, label='p+v (t=1)')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-5,8); plt.ylim(-5,8)
plt.legend(); plt.title('$\\vec{p} + t\\vec{v}$ — line shifted from origin')
plt.show()
```

---
## 3. Line Through Two Points

Given two points $\vec{a}$ and $\vec{b}$, the direction is $\vec{v} = \vec{b} - \vec{a}$:

$$\text{Line}(t) = \vec{a} + t(\vec{b} - \vec{a})$$

$t=0$ → $\vec{a}$, $t=1$ → $\vec{b}$, $t=0.5$ → midpoint

```python
a = np.array([1, 1]); b = np.array([5, 3])
v = b - a

print(f"Direction v = b - a = {v}")
print(f"Midpoint (t=0.5): {a + 0.5*v}")

t_values = np.linspace(-1, 2, 100)
points = np.array([a + t*v for t in t_values])

plt.figure(figsize=(6,6))
plt.plot(points[:,0], points[:,1], 'b-', label='Line')
plt.axhline(0, color='gray', linewidth=0.5); plt.axvline(0, color='gray', linewidth=0.5)
plt.scatter(a[0], a[1], color='green', s=150, label='a (t=0)')
plt.scatter(b[0], b[1], color='red', s=150, label='b (t=1)')
plt.scatter(*(a + 0.5*v), color='orange', s=150, label='Midpoint (t=0.5)')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-2,8); plt.ylim(-2,6)
plt.legend(); plt.title('Line through two points: $\\vec{a} + t(\\vec{b}-\\vec{a})$')
plt.show()
```

### 🤔 Think About This

**Before running the next cell:** If you have $\vec{a} = \begin{bmatrix}2 \\ 3\end{bmatrix}$ and $\vec{b} = \begin{bmatrix}8 \\ 6\end{bmatrix}$, what is the direction vector? What is the midpoint?

```python
a = np.array([2, 3]); b = np.array([8, 6])
v = b - a
mid = a + 0.5*v

print(f"Direction: {v}")
print(f"Midpoint: {mid}")
print(f"Check: (a + b)/2 = {(a+b)/2} ← same thing!")
```

## 🤖 ML Connection

### Decision Boundaries

In classification, a **linear decision boundary** separates classes. It's a line (or hyperplane):

$$w_1 x_1 + w_2 x_2 + b = 0$$

You can represent it parametrically as a line through two points on the boundary.

### Gradient Descent

$$\theta_{\text{new}} = \theta_{\text{old}} - \alpha \nabla L$$

That's $\vec{p}_{t+1} = \vec{p}_t + t \cdot \vec{v}$ where $t = -\alpha$ and $\vec{v} = \nabla L$.

### Linear Interpolation (Data Augmentation)

Given two data points, $t=0.5$ gives their **average** — used in data augmentation, image morphing, and GAN latent space interpolation.

## ✏️ Exercise 1

$\vec{a} = \begin{bmatrix}2 \\ 3\end{bmatrix}$, $\vec{b} = \begin{bmatrix}6 \\ 5\end{bmatrix}$

1. Direction $\vec{v} = ?$  
2. Point at $t = 0.25$?  
3. Point at $t = -1$?  
4. Plot everything.

```python
a = np.array([2,3]); b = np.array([6,5])
v = b - a
print(f"1. Direction: {v}")
print(f"2. t=0.25: {a + 0.25*v}")
print(f"3. t=-1: {a + (-1)*v}")

t_vals = np.linspace(-2, 3, 100)
points = np.array([a + t*v for t in t_vals])

plt.figure(figsize=(6,6))
plt.plot(points[:,0], points[:,1], 'b-', label='Line')
plt.axhline(0, color='gray', linewidth=0.5); plt.axvline(0, color='gray', linewidth=0.5)
plt.scatter(*a, color='green', s=150, label='a (t=0)')
plt.scatter(*b, color='red', s=150, label='b (t=1)')
plt.scatter(*(a+0.25*v), color='orange', s=150, label='t=0.25')
plt.scatter(*(a-v), color='purple', s=150, label='t=-1')
plt.grid(alpha=0.3); plt.axis('equal'); plt.legend()
plt.title('Exercise: Line through a and b')
plt.show()
```

## ✏️ Exercise 2 — 3D Line

$\vec{p} = \begin{bmatrix}1 \\ 2 \\ 3\end{bmatrix}$, $\vec{v} = \begin{bmatrix}2 \\ -1 \\ 0\end{bmatrix}$

1. Find points at $t = 0, 2, -1$  
2. Verify they are colinear

```python
p = np.array([1,2,3]); v = np.array([2,-1,0])
t0 = p + 0*v; t2 = p + 2*v; tn1 = p + (-1)*v

print(f"t=0:  {t0}")
print(f"t=2:  {t2}")
print(f"t=-1: {tn1}")
d1 = t2 - t0; d2 = tn1 - t0
print(f"Colinear? {np.allclose(d1, -2*d2)} ✅")
```
