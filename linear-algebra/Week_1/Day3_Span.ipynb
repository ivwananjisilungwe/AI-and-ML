## 🧠 Intuition: What is a Span?

The **span** of a set of vectors is **all possible points you can reach** using linear combinations of them.

### One Vector → A Line

Given $\vec{v} = \begin{bmatrix}1 \\ 2\end{bmatrix}$, the span is all points $c \cdot \vec{v}$ for any scalar $c$.

That's a **line** through the origin.

### Two Vectors (different directions) → A Plane

Given $\vec{a}$ and $\vec{b}$ pointing in **different directions**:  
Their span is the **entire 2D plane**. You can reach any point $(x, y)$.

### Two Vectors (same direction) → Still Just A Line

If $\vec{b} = k \cdot \vec{a}$ (one is a scaled version of the other), their span is **still just a line**.

### Three Vectors in 3D

- All random: span = the **entire 3D space**  
- One lies in the plane of the other two: span = a **plane** (2D subspace)  
- All on the same line: span = a **line** (1D subspace)  
- All zero: span = the **origin** (0D subspace)

> **Key insight:** The span is the set of all destinations you can reach by scaling and adding your available vectors.

### 🤖 ML Connection

Your **feature space** is a span:
- 2 features → data lives in a 2D plane  
- 100 features → 100D space (can't visualize, but the math is the same)  
- Often your data only occupies a **lower-dimensional subspace** → this is what **PCA** detects.

In neural networks, each layer **expands or contracts the span** of the data.

## 🔬 Visualizing Span in 2D

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Span of One Vector = A Line

**Before running: what shape do you expect to see?**

```python
v = np.array([1, 2])
scalars = np.random.uniform(-3, 3, 500)
points = np.array([c * v for c in scalars])

plt.figure(figsize=(6, 6))
plt.scatter(points[:, 0], points[:, 1], alpha=0.5, s=10)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.arrow(0, 0, v[0], v[1], head_width=0.2, head_length=0.3, color='red', width=0.05)
plt.grid(alpha=0.3)
plt.axis('equal')
plt.xlim(-5, 5); plt.ylim(-5, 5)
plt.title('Span of one vector = a LINE')
plt.show()
print("✅ All points lie on one line through the origin.")
```

### Span of Two Vectors (Different Directions) = A Plane

**Before running: what do you expect with [1,0] and [0,1]?**

```python
a = np.array([1, 0]); b = np.array([0, 1])
c1s = np.random.uniform(-3, 3, 1000)
c2s = np.random.uniform(-3, 3, 1000)
points = np.array([c1*a + c2*b for c1,c2 in zip(c1s,c2s)])

plt.figure(figsize=(6, 6))
plt.scatter(points[:, 0], points[:, 1], alpha=0.3, s=5)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.arrow(0,0,a[0],a[1], head_width=0.15,head_length=0.2, color='red', width=0.03, label='a')
plt.arrow(0,0,b[0],b[1], head_width=0.15,head_length=0.2, color='blue', width=0.03, label='b')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-4,4); plt.ylim(-4,4)
plt.title('Span of [1,0] and [0,1] = ENTIRE 2D plane')
plt.legend(); plt.show()
print("✅ Points fill the whole plane — any (x,y) is reachable.")
```

---
### 🤔 Think Before You Run

What if $\vec{b} = 2 \cdot \vec{a}$ (they're colinear)?

**Question:** Can you reach the point $(3, 5)$ using $\vec{a} = \begin{bmatrix}1 \\ 1\end{bmatrix}$ and $\vec{b} = \begin{bmatrix}2 \\ 2\end{bmatrix}$?

> *Think about it. Both vectors lie on the line y=x. Any combination stays on y=x. Point (3,5) is NOT on y=x. So...*

```python
a = np.array([1, 1]); b = np.array([2, 4])  # b = 2*a

c1s = np.random.uniform(-3, 3, 1000)
c2s = np.random.uniform(-3, 3, 1000)
points = np.array([c1*a + c2*b for c1,c2 in zip(c1s,c2s)])

plt.figure(figsize=(6, 6))
plt.scatter(points[:,0], points[:,1], alpha=0.3, s=5)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.scatter(3, 5, color='red', s=150, label='Target (3,5) — unreachable!')
plt.arrow(0,0,a[0],a[1], head_width=0.15,head_length=0.2, color='red', width=0.03, label='a')
plt.arrow(0,0,b[0],b[1], head_width=0.15,head_length=0.2, color='blue', width=0.03, label='b')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-5,8); plt.ylim(-5,8)
plt.title('Colinear vectors = span is still JUST a line')
plt.legend(); plt.show()
print("❌ (3,5) is not on the line — no combination of colinear vectors can reach it.")
```

## ✏️ Exercise 1

Given $\vec{a} = \begin{bmatrix}2 \\ 0\end{bmatrix}$ and $\vec{b} = \begin{bmatrix}0 \\ 3\end{bmatrix}$:

1. What shape is their span?  
2. Generate 2000 random combinations and plot them.  
3. Can you reach $(4, 6)$? What coefficients $(c_1, c_2)$ do it?

```python
a = np.array([2, 0]); b = np.array([0, 3])
c1s = np.random.uniform(-3, 3, 2000)
c2s = np.random.uniform(-3, 3, 2000)
points = np.array([c1*a + c2*b for c1,c2 in zip(c1s,c2s)])

plt.figure(figsize=(6,6))
plt.scatter(points[:,0], points[:,1], alpha=0.2, s=3)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.arrow(0,0,a[0],a[1], head_width=0.15,head_length=0.2, color='red', width=0.03, label='a')
plt.arrow(0,0,b[0],b[1], head_width=0.15,head_length=0.2, color='blue', width=0.03, label='b')

target = 2*a + 2*b
plt.scatter(target[0], target[1], color='green', s=150, label=f'(4,6): 2a+2b')
plt.grid(alpha=0.3); plt.axis('equal'); plt.xlim(-8,8); plt.ylim(-8,8)
plt.legend(); plt.title('Span = entire 2D plane (rectangular grid)')
plt.show()
print(f"(4,6) reached with c1=2, c2=2: {2*a + 2*b}")
```
