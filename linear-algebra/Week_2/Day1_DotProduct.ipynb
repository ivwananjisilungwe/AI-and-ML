## 🧠 Intuition: What is a Dot Product?

The dot product is a way to **multiply two vectors** to get a **single number** (a scalar).

### The Formula

$$\vec{a} \cdot \vec{b} = a_1 b_1 + a_2 b_2 + \dots + a_n b_n = \sum_{i=1}^{n} a_i b_i$$

Multiply corresponding elements, then **add them all up**.

### Intuition: "How much do they align?"

- Large positive dot product → vectors point in **similar directions**  
- Zero dot product → vectors are **perpendicular** (orthogonal)  
- Large negative dot product → vectors point in **opposite directions**

### Visual

```
a = [1, 2]    b = [3, 4]

a · b = 1×3 + 2×4 = 3 + 8 = 11

  Step 1: multiply    Step 2: add
  [1,2] · [3,4]      [1×3, 2×4]
       ↓                  ↓
  [3, 8]               11
```

### 🤖 Why This Matters for ML

**Every neural network layer is a dot product:**

$$z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b = \vec{w} \cdot \vec{x} + b$$

The dot product measures how well the input $\vec{x}$ matches the weights $\vec{w}$.  
This is the fundamental operation in:
- Linear regression
- Neural network forward pass
- Attention mechanisms (transformers)
- Cosine similarity (search engines)

## 🔢 Computing the Dot Product in NumPy

````python
import numpy as np

a = np.array([1, 2])
b = np.array([3, 4])

# Method 1: Manual (step by step)
step1 = a * b          # Element-wise multiply: [3, 8]
result_manual = np.sum(step1)  # Sum: 11

print("a =", a)
print("b =", b)
print()
print("Step 1 — a * b     =", step1)
print("Step 2 — np.sum(...) =", result_manual)
print()

# Method 2: NumPy built-in
result_builtin = np.dot(a, b)
print("np.dot(a, b)       =", result_builtin)
print()

# Method 3: @ operator (Python 3.5+)
result_at = a @ b
print("a @ b              =", result_at)
print()

print(f"All match? {result_manual == result_builtin == result_at} ✅")
````

### Properties

1. **Commutative:** $\vec{a} \cdot \vec{b} = \vec{b} \cdot \vec{a}$  
2. **Distributive:** $\vec{a} \cdot (\vec{b} + \vec{c}) = \vec{a} \cdot \vec{b} + \vec{a} \cdot \vec{c}$  
3. **Scalar multiplication:** $(c\vec{a}) \cdot \vec{b} = c(\vec{a} \cdot \vec{b})$  
4. **Dot with itself:** $\vec{a} \cdot \vec{a} = \|\vec{a}\|^2$ (the squared length)

````python
a = np.array([1, 2, 3]); b = np.array([4, 5, 6]); c = np.array([0, -1, 2])

# Commutative
print(f"a·b = {np.dot(a,b)}, b·a = {np.dot(b,a)} — same ✅")

# Distributive
lhs = np.dot(a, b + c)
rhs = np.dot(a, b) + np.dot(a, c)
print(f"a·(b+c) = {lhs}, a·b + a·c = {rhs} — distributive holds ✅")

# Dot with itself = squared length
dot_self = np.dot(a, a)
norm_sq = np.linalg.norm(a) ** 2
print(f"a·a = {dot_self}, ||a||² = {norm_sq:.2f} — match ✅")
````

## ✏️ Exercise 1

Compute the dot product of these vectors in **three ways** (manual, `np.dot`, `@`):

$$\vec{x} = \begin{bmatrix}2 & -3 & 1\end{bmatrix}, \quad \vec{y} = \begin{bmatrix}4 & 2 & -5\end{bmatrix}$$

````python
x = np.array([2, -3, 1])
y = np.array([4, 2, -5])

manual = np.sum(x * y)
builtin = np.dot(x, y)
at = x @ y

print(f"x · y = {manual}")
print(f"np.dot = {builtin}")
print(f"x @ y  = {at}")
````

## ✏️ Exercise 2 — ML Context

You have 3 features and a linear model:

$$\text{score} = \vec{w} \cdot \vec{x} + b$$

Features: $\vec{x} = \begin{bmatrix}2.5 & 30 & 1\end{bmatrix}$ (hours studied, age, previous course)  
Weights: $\vec{w} = \begin{bmatrix}8 & 0.5 & 10\end{bmatrix}$  
Bias: $b = 5$

Use a **dot product** to compute the predicted score.

````python
x = np.array([2.5, 30, 1])
w = np.array([8, 0.5, 10])
b = 5

score = np.dot(w, x) + b
print(f"Predicted score: {score:.2f}")
````
