## 🧠 Intuition: The Geometric Dot Product

### Two Equivalent Formulas

$$\vec{a} \cdot \vec{b} = \sum a_i b_i = \|\vec{a}\| \, \|\vec{b}\| \, \cos \theta$$

The second formula reveals the **geometry**:

$$\vec{a} \cdot \vec{b} = (\text{length of } \vec{a}) \times (\text{length of } \vec{b}) \times \cos(\text{angle between them})$$

### What is Projection?

Imagine shining a flashlight directly down onto vector $\vec{b}$:  
The **shadow** that $\vec{a}$ casts on $\vec{b}$ is its **projection**.

```
         a
         ↑
        /|\
       / | \
      /  |  \     ← light shining down
     /   |   \
    /    |    \
   └─────┴─────→ b
    ←-- proj --→
```

The projection answers: **"How much of a points in the direction of b?"**

### The Formula

$$\text{proj}_{\vec{b}}(\vec{a}) = \frac{\vec{a} \cdot \vec{b}}{\vec{b} \cdot \vec{b}} \; \vec{b}$$

The scalar $\frac{\vec{a} \cdot \vec{b}}{\vec{b} \cdot \vec{b}}$ is the **coefficient** — how many steps of $\vec{b}$ we need.

- A) A vector equal to $\vec{a}$  
- B) The zero vector  
- C) A vector equal to $\vec{b}$

> *(Answer: B — if dot = 0, vectors are orthogonal, so a has zero shadow on b)*

## 🔢 Computing Projections in NumPy

````python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([4, 3])
b = np.array([5, 0])

# Project a onto b
coeff = np.dot(a, b) / np.dot(b, b)
proj = coeff * b

print(f"a = {a}")
print(f"b = {b}")
print(f"Coefficient (a·b)/(b·b) = {coeff:.2f}")
print(f"Projection = {coeff:.2f} * {b} = {proj}")
print()

# Verify: residual is orthogonal to b
residual = a - proj
print(f"Residual (a - proj) = {residual}")
print(f"residual · b = {np.dot(residual, b):.2f}  ← should be ~0 ✅")
````

### Visualizing the Projection

````python
def plot_projection(a, b):
    coeff = np.dot(a, b) / np.dot(b, b)
    proj = coeff * b
    
    plt.figure(figsize=(6, 6))
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    
    # Original vectors
    plt.arrow(0, 0, a[0], a[1], head_width=0.2, head_length=0.3, color='red', width=0.05, label='a')
    plt.arrow(0, 0, b[0], b[1], head_width=0.2, head_length=0.3, color='blue', width=0.05, label='b')
    
    # Projection
    plt.arrow(0, 0, proj[0], proj[1], head_width=0.15, head_length=0.2, color='green', width=0.05, label=f'proj (len={np.linalg.norm(proj):.1f})')
    
    # Residual (dashed)
    plt.plot([a[0], proj[0]], [a[1], proj[1]], 'k--', alpha=0.5, label='residual')
    
    plt.grid(alpha=0.3)
    plt.axis('equal')
    plt.legend()
    plt.title(f'Projection of a onto b')
    plt.show()

plot_projection(np.array([4, 3]), np.array([5, 0]))
````

### Try Different Angles

Let's see how the projection changes as we rotate $\vec{a}$.

````python
b = np.array([5, 0])

# Try a at different angles
for angle, label in [(0, 'same direction'), (30, '30°'), (60, '60°'), (90, 'orthogonal'), (120, 'obtuse')]:
    theta = np.radians(angle)
    a = np.array([4 * np.cos(theta), 4 * np.sin(theta)])
    coeff = np.dot(a, b) / np.dot(b, b)
    proj = coeff * b
    print(f"{label:15s}  a={a}  coeff={coeff:.2f}  |proj|={np.linalg.norm(proj):.2f}")
````

## 🤖 ML Connection

### Attention Mechanism

In Transformer models (GPT, BERT, etc.), the **attention score** between two words is a dot product:

$$\text{score} = \vec{q} \cdot \vec{k}$$

This measures how much the query vector $\vec{q}$ aligns with the key vector $\vec{k}$ — exactly like a projection.

### Feature Contribution

In linear models, $w_i x_i$ tells you how much feature $i$ contributes to the prediction. That's a **projection** of the feature onto the weight direction.

### Least Squares

The best-fit line in regression projects the target vector onto the column space of the design matrix — a higher-dimensional projection. We'll see this in Week 12.

Given $\vec{a} = \begin{bmatrix}3 \\ 4\end{bmatrix}$ and $\vec{b} = \begin{bmatrix}2 \\ 0\end{bmatrix}$:

1. Compute the projection of $\vec{a}$ onto $\vec{b}$  
2. Compute the residual $\vec{a} - \text{proj}$  
3. Verify the residual is orthogonal to $\vec{b}$  
4. Plot

````python
a = np.array([3, 4])
b = np.array([2, 0])

# Projection
coeff = np.dot(a, b) / np.dot(b, b)
proj = coeff * b
residual = a - proj

print(f"Projection: {proj}")
print(f"Residual: {residual}")
print(f"Residual · b = {np.dot(residual, b):.2f} (should be 0)")

plot_projection(a, b)
````
