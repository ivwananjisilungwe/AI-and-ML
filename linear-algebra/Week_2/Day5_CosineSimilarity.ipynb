## 🧠 Intuition: Cosine Similarity

**Cosine similarity** measures how similar two vectors are — based on direction, ignoring magnitude.

### The Formula

$$\text{cosine\_similarity}(\vec{a}, \vec{b}) = \frac{\vec{a} \cdot \vec{b}}{\|\vec{a}\| \, \|\vec{b}\|} = \cos\theta$$

### The Scale

| Value | Meaning |
|-------|---------|
| $+1$ | Same direction (identical) |
| $0$ | Orthogonal (unrelated) |
| $-1$ | Opposite direction |

```
   -1          0         +1
    ├──────────┼──────────┤
  opposite   unrelated   same
```

### Why This is Everywhere in ML

Cosine similarity is **scale-invariant**. Whether an embedding vector has length 5 or 500, cosine similarity only cares about direction.

This is crucial because:
- Word embeddings from different contexts may have different scales
- User/item embeddings in recommendation systems are compared by direction
- Document vectors (TF-IDF) of different lengths need fair comparison

## 🔢 Computing Cosine Similarity

````python
import numpy as np

def cosine_similarity(a, b):
    dot = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return np.clip(dot / (norm_a * norm_b), -1.0, 1.0)

# Test the extremes
a = np.array([5, 0])
b_same = np.array([5, 0])
b_orth = np.array([0, 7])
b_opp = np.array([-5, 0])
b_partial = np.array([3, 4])

print(f"Same direction: {cosine_similarity(a, b_same):.2f}  → +1 ✅")
print(f"Orthogonal:      {cosine_similarity(a, b_orth):.2f}  →  0 🟢")
print(f"Opposite:        {cosine_similarity(a, b_opp):.2f}  → -1 ✅")
print(f"45° angle:       {cosine_similarity(a, b_partial):.2f}  →  0.6 (cos 53°) ≈ 0.6")
````

## 🤖 Real ML: Word Embedding Similarity

Let's simulate word embeddings (word2vec-style) and measure semantic similarity.

````python
np.random.seed(42)

# Simulated 50-dimensional word embeddings
words = {
    'king':   np.random.randn(50),
    'queen':  np.random.randn(50),
    'man':    np.random.randn(50),
    'woman':  np.random.randn(50),
    'apple':  np.random.randn(50),
    'orange': np.random.randn(50),
    'car':    np.random.randn(50),
    'road':   np.random.randn(50),
}

# Compute all pairwise similarities
word_list = list(words.keys())
print("Cosine Similarity Matrix:\n")
print(f"{'':>8}", end='')
for w2 in word_list:
    print(f"{w2:>8}", end='')
print()

for w1 in word_list:
    print(f"{w1:>8}", end='')
    for w2 in word_list:
        sim = cosine_similarity(words[w1], words[w2])
        print(f"{sim:>8.3f}", end='')
    print()
````

## 🤖 ML Applications

### 1. Information Retrieval (Search)

Search query "machine learning tutorial" → embed into vector → find documents with highest cosine similarity.

### 2. Recommendation Systems

User embedding $\vec{u}$ and item embedding $\vec{v}$: $\cos(\vec{u}, \vec{v})$ > 0.8 → recommend.

### 3. Semantic Similarity (NLP)

"king" vs "queen" have higher cosine similarity than "king" vs "car" in good embeddings.

### 4. Clustering

K-means with cosine distance $(1 - \cos\theta)$ groups items by direction — the standard approach for high-dimensional text data.

## ✏️ Exercise 1: Document Similarity

Three documents with TF-IDF vectors:
- Doc1: $\begin{bmatrix}1 & 2 & 3\end{bmatrix}$ (sports news)  
- Doc2: $\begin{bmatrix}2 & 4 & 6\end{bmatrix}$ (more sports news)  
- Doc3: $\begin{bmatrix}3 & 1 & 0\end{bmatrix}$ (tech news)

1. Compute cosine similarity between Doc1 and Doc2  
2. Compute cosine similarity between Doc1 and Doc3  
3. Which pair is more similar?  
4. **Think first:** Is Doc1 and Doc2 identical from a direction perspective?

````python
d1 = np.array([1, 2, 3])
d2 = np.array([2, 4, 6])
d3 = np.array([3, 1, 0])

sim_12 = cosine_similarity(d1, d2)
sim_13 = cosine_similarity(d1, d3)

print(f"cos(Doc1, Doc2) = {sim_12:.4f}")
print(f"cos(Doc1, Doc3) = {sim_13:.4f}")
print()
print(f"'Sports' docs (1 & 2) are {sim_12:.0%} directionally identical!")
print(f"'Sports' vs 'Tech' docs (1 & 3) have cos = {sim_13:.2f}")
````
