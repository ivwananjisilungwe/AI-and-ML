## 🧠 Intuition: What is a Norm?

The **norm** (also called **magnitude** or **length**) of a vector is its size.

### The Formula

$$\|\vec{v}\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2} = \sqrt{\sum_{i=1}^{n} v_i^2}$$

This is just the **Pythagorean theorem** extended to $n$ dimensions.

### Connection to Dot Product

$$\|\vec{v}\| = \sqrt{\vec{v} \cdot \vec{v}}$$

The norm is the square root of the dot product of a vector with itself.

### What is a Unit Vector?

A **unit vector** has length exactly 1. You create one by dividing a vector by its norm:

$$\hat{v} = \frac{\vec{v}}{\|\vec{v}\|}$$

This is called **normalizing** the vector — it keeps the direction but sets length = 1.

```
   Before (length = 5)     After (length = 1)
        ↑                       ↑
        |                       |
        |    v                  |    v̂
        |                       |
        |                       |
   Same direction, different length
```

### 🤖 Why This Matters for ML

**Feature normalization:** When features have different scales (age vs. income), models can be biased.  
We normalize each feature vector so all features have the same scale.

**Weight decay:** Regularization penalizes the norm of weight vectors to prevent overfitting.

**Embeddings:** Word embeddings are often normalized so cosine similarity measures only direction.

## 🔢 Computing Norms in NumPy

````python
import numpy as np

v = np.array([3, 4])

# Method 1: Manual (step by step)
squared = v ** 2          # [9, 16]
summed = np.sum(squared)  # 25
manual = np.sqrt(summed)  # 5.0

print("Manual computation:")
print(f"  v**2       = {squared}")
print(f"  np.sum(...) = {summed}")
print(f"  np.sqrt(...)= {manual}")
print()

# Method 2: Via dot product
norm_from_dot = np.sqrt(np.dot(v, v))
print(f"sqrt(v·v)     = {norm_from_dot}")
print()

# Method 3: Built-in
norm = np.linalg.norm(v)
print(f"np.linalg.norm(v) = {norm}")
print()

print(f"All match? {manual == norm_from_dot == norm} ✅")
````

### Creating Unit Vectors

````python
v = np.array([3, 4])
norm = np.linalg.norm(v)
unit_v = v / norm

print(f"v        = {v}")
print(f"||v||    = {norm}")
print(f"unit_v   = {unit_v}")
print(f"||unit_v|| = {np.linalg.norm(unit_v):.6f}  ← should be 1 ✅")
print()

# Direction is preserved
print(f"v / ||v|| = {v / norm}")
print("Same direction, different length.")
````

### 🔍 Discover the Pattern

**Before running:** What happens to the norm when you scale a vector?

If $\|\vec{v}\| = 5$, what is $\|2\vec{v}\|$? What is $\|0.5\vec{v}\|$? What is $\|-3\vec{v}\|$?

````python
v = np.array([3, 4])
print(f"||v||     = {np.linalg.norm(v)}")
print(f"||2*v||   = {np.linalg.norm(2*v)}")
print(f"||0.5*v|| = {np.linalg.norm(0.5*v)}")
print(f"||-3*v||  = {np.linalg.norm(-3*v)}")
print()
print("Pattern: ||c * v|| = |c| * ||v||")
print("Scaling the vector scales its length by the same factor.")
````

## 🤖 ML Connection: Normalization

### Feature Scaling

Suppose two customers:
- Customer A: $\begin{bmatrix}20 & 40 & 60\end{bmatrix}$
- Customer B: $\begin{bmatrix}200 & 400 & 600\end{bmatrix}$

Same **pattern** (direction), different **scale** (magnitude).  
Normalizing (dividing by norm) makes them comparable.

### L2 Regularization (Weight Decay)

$$\text{Loss} = \text{error} + \lambda \|\vec{w}\|^2$$

Penalizing the norm of weights prevents them from growing too large — reducing overfitting.

### Normalized Embeddings

Word2Vec, GloVe, and BERT embeddings are often normalized so cosine similarity (Day 5) measures pure direction, not magnitude.

1. Compute the norm of $\vec{v} = \begin{bmatrix}6 & 8\end{bmatrix}$ using all three methods  
2. Create a unit vector from $\vec{v}$  
3. Verify its length is 1

````python
v = np.array([6, 8])

# Norms
manual = np.sqrt(np.sum(v**2))
from_dot = np.sqrt(np.dot(v, v))
builtin = np.linalg.norm(v)
print(f"Manual: {manual}, From dot: {from_dot}, Built-in: {builtin}")

# Unit vector
unit = v / builtin
print(f"Unit vector: {unit}")
print(f"Length of unit vector: {np.linalg.norm(unit):.6f}")
````
