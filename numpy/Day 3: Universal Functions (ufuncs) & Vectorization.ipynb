# 1. What is a Ufunc?

A Universal Function (ufunc) is a function that operates on every element of an array.

Notice:
- No loops
- No indexing
- Applied automatically to every element

````python
# Example: np.sqrt() applies to every element automatically
arr = np.array([1, 2, 3, 4])

print(np.sqrt(arr))
````

# 2. Element-Wise Operations

NumPy supports all basic arithmetic as ufuncs.

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

# Addition
print("a + b =", a + b)
print("np.add(a,b) =", np.add(a, b))

# Subtraction
print("np.subtract(b, a) =", np.subtract(b, a))
print()

# Multiplication
print("np.multiply(a, b) =", np.multiply(a, b))
print()print()


# Division
print("np.divide(b, a) =", np.divide(b, a))
print()

# Power
print("np.power(a, 2) =", np.power(a, 2))
````

# 3. Square Root

````python
arr = np.array([1, 4, 9, 16])

print(np.sqrt(arr))
````

# 4. Exponential

# 6. Trigonometric Functions

NumPy provides sin, cos, tan and more.

````python
angles = np.array([0, np.pi/2, np.pi])

print("np.sin(angles) =", np.sin(angles))
print("np.cos(angles) =", np.cos(angles))
````

# 7. Rounding Functions

- `np.round()` → standard rounding
- `np.floor()` → always rounds down
- `np.ceil()` → always rounds up

````python
arr = np.array([1.2, 1.7, 3.4, 3.9])

print("np.round(arr) =", np.round(arr))
print("np.floor(arr) =", np.floor(arr))
print("np.ceil(arr)  =", np.ceil(arr))
````

# 8. np.clip()

Limits values to a range.

Signature: `np.clip(arr, min, max)`

````python
scores = np.array([40, 60, 80, 120])

print(np.clip(scores, 0, 100))
````

# 9. np.where()

Conditional selection.

Signature: `np.where(condition, x, y)`

````python
scores = np.array([45, 78, 90, 30])

result = np.where(scores >= 50, "Pass", "Fail")

print(result)
````

# 10. Vectorization

Bad (slow Python loop):

```python
result = []
for x in scores:
    result.append(x + 5)
```

Good (vectorized NumPy):

```python
result = scores + 5
```

This is **Vectorization**.

## Why Vectorization Matters

- Loop version: Python iterates slowly
- NumPy version: executes optimized C code internally
- Much faster for large arrays

````python
# Vectorization example
scores = np.array([45, 78, 90, 30])

# Bad way (don't do this)
result = []
for x in scores:
    result.append(x + 5)
print("Loop result:", result)

# Good way (vectorized)
result = scores + 5
print("Vectorized:", result)
````

````python
# Speed comparison: NumPy vs Python loop
import time

big = np.arange(1_000_000)

# Python loop version
start = time.time()
result = []
for x in big:
    result.append(x * 2)
loop_time = time.time() - start

# NumPy vectorized version
start = time.time()
result = big * 2
numpy_time = time.time() - start

print(f"Python loop: {loop_time:.4f} seconds")
print(f"NumPy:       {numpy_time:.4f} seconds")
print(f"NumPy is {loop_time / numpy_time:.1f}x faster!")
````

# 11. np.vectorize()

Allows normal Python functions to work on arrays.

Note: Convenient, but not actually faster than a loop.

````python
def square(x):
    return x * x

v_square = np.vectorize(square)

print(v_square(np.array([1, 2, 3, 4])))
````

# Mini Project: Smart Grade Processor

````python
grades = np.array([
    45, 55, 72, 88, 91,
    33, 67, 79, 100, 58
])

# Task 1: Add 5 bonus marks to every student
bonus = grades + 5
print("After bonus:", bonus)
print()

# Task 2: Calculate square root of all grades
print("Square roots:", np.sqrt(grades))
print()

# Task 3: Create Pass/Fail labels (Pass >= 50, Fail < 50)
labels = np.where(grades >= 50, "Pass", "Fail")
print("Pass/Fail:", labels)
print()

# Task 4: Round all grades after multiplying by 1.15
scaled = np.round(grades * 1.15)
print("After scaling:", scaled)
print()

# Task 5: Clip grades to maximum 100
clipped = np.clip(grades, 0, 100)
print("Clipped to 100:", clipped)
````

````python
# Task 6: Create a custom letter_grade function and apply with np.vectorize()

def letter_grade(x):
    if x >= 90:
        return "A"
    elif x >= 80:
        return "B"
    elif x >= 70:
        return "C"
    elif x >= 60:
        return "D"
    else:
        return "F"

v_letter_grade = np.vectorize(letter_grade)

print("Grades:", grades)
print("Letters:", v_letter_grade(grades))
````
