# 1. What are Aggregations?

An aggregation reduces many values into one result.

Example:

```python
scores = np.array([70, 80, 90, 100])
```

Instead of keeping:

`70 80 90 100`

we summarize:

- Average = 85
- Maximum = 100
- Minimum = 70

# 2. np.sum()

Adds all elements.

Signature: `np.sum(a, axis=None, dtype=None, keepdims=False)`

| Parameter | Meaning |
|-----------|---------|
| a | Input array |
| axis | Direction to calculate |
| dtype | Output type |
| keepdims | Keep dimensions |

````python
# 1D example
numbers = np.array([10, 20, 30, 40])

print(np.sum(numbers))
````

````python
# 2D example
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(np.sum(arr))
````

# 3. Understanding axis

This is one of the most important NumPy concepts.

Given:

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

Visual:

```
       axis=1 →
       
       1 2 3
axis=0 4 5 6
   ↓
```

**axis=0** → Operate down rows (columns).

**axis=1** → Operate across columns (rows).

````python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Sum axis=0 (column-wise):", np.sum(arr, axis=0))
print("Sum axis=1 (row-wise):", np.sum(arr, axis=1))
````

# 4. np.mean()

Calculates average.

Signature: `np.mean(a, axis=None)`

Formula: (sum of values) / number of values

````python
scores = np.array([70, 80, 90, 100])

print(np.mean(scores))
````

# 5. np.median()

Returns the middle value.

Signature: `np.median(a, axis=None)`

````python
# Odd number of values
numbers = np.array([1, 3, 5, 7, 9])

print(np.median(numbers))
````

````python
# Even number of values
numbers = np.array([1, 2, 3, 4])

print(np.median(numbers))

# Because (2 + 3) / 2 = 2.5
````

# 6. np.std()

Standard deviation. Measures how spread out data is.

Signature: `np.std(a, axis=None)`

- Small standard deviation → Values are close together
- Large standard deviation → Values are spread apart

````python
data = np.array([10, 20, 30])

print(np.std(data))
````

# 7. np.var()

Variance. Variance is: standard deviation²

Signature: `np.var(a, axis=None)`

Relationship: `variance = std × std`

````python
data = np.array([10, 20, 30])

print("Variance:", np.var(data))
print("Std squared:", np.std(data) ** 2)
````

# 8. np.min()

Find minimum value.

Signature: `np.min(a, axis=None)`

````python
scores = np.array([50, 80, 30, 90])

print(np.min(scores))
````

# 9. np.max()

Find maximum value.

Signature: `np.max(a, axis=None)`

````python
scores = np.array([50, 80, 30, 90])

print(np.max(scores))
````

# 10. np.argmin()

Returns the index of the smallest value.

Signature: `np.argmin(a, axis=None)`

````python
scores = np.array([50, 80, 30, 90])

print(np.argmin(scores))

# Index:
# 0 → 50
# 1 → 80
# 2 → 30  ← smallest
# 3 → 90
````

# 11. np.argmax()

Returns index of largest value.

Signature: `np.argmax(a, axis=None)`

````python
scores = np.array([50, 80, 30, 90])

print(np.argmax(scores))
````

# 12. np.percentile()

Finds values below which a percentage of data falls.

Signature: `np.percentile(a, q, axis=None)`

Common percentiles:

| Percentile | Meaning |
|------------|---------|
| 25 | Q1 |
| 50 | Median |
| 75 | Q3 |
| 100 | Maximum |

````python
scores = np.array([50, 60, 70, 80, 90, 100])

print("25th percentile:", np.percentile(scores, 25))
print("50th percentile:", np.percentile(scores, 50))
print("75th percentile:", np.percentile(scores, 75))
````

# 13. np.quantile()

Similar to percentile but uses 0-1 scale.

Signature: `np.quantile(a, q)`

Comparison:

- `np.percentile(data, 50)`
- `np.quantile(data, 0.5)`

Same result.

````python
data = np.array([50, 60, 70, 80, 90, 100])

print("Percentile 50:", np.percentile(data, 50))
print("Quantile 0.5:", np.quantile(data, 0.5))
````

# 14. np.cumsum()

Cumulative sum. Each element becomes the sum of previous values.

Signature: `np.cumsum(a, axis=None)`

````python
sales = np.array([100, 200, 300])

print(np.cumsum(sales))

# Explanation:
# 100
# 100 + 200 = 300
# 100 + 200 + 300 = 600
````

# 15. np.cumprod()

Cumulative multiplication.

Signature: `np.cumprod(a, axis=None)`

````python
numbers = np.array([2, 3, 4])

print(np.cumprod(numbers))

# 2
# 2 * 3 = 6
# 2 * 3 * 4 = 24
````

# 16. np.any()

Checks if at least one value is True.

Signature: `np.any(a, axis=None)`

````python
scores = np.array([40, 60, 80])

print(np.any(scores < 50))

# True because 40 exists
````

# 17. np.all()

Checks if all values are True.

Signature: `np.all(a, axis=None)`

````python
scores = np.array([60, 70, 80])

print(np.all(scores >= 50))

# True because all scores are >= 50
````

# 18. Handling Missing Data

Missing values are represented as: `np.nan`

Problem: NaN contaminates calculations.

````python
data = np.array([10, 20, np.nan, 40])

print(np.mean(data))

# Output: nan (because NaN contaminates)
````

# 19. np.nanmean()

Ignores NaN values.

Signature: `np.nanmean(a)`

Calculation: (10 + 20 + 40) / 3 = 23.333

````python
data = np.array([10, 20, np.nan, 40])

print(np.nanmean(data))
````

# 20. np.nansum()

Sum ignoring NaN.

Signature: `np.nansum(a)`

````python
data = np.array([10, 20, np.nan, 40])

print(np.nansum(data))
````

# Mini Project: Exam Statistics Dashboard

````python
scores = np.array([
    45, 55, 67, 72, 88,
    91, 34, 76, 82, 95,
    60, 70
])

# Calculate:
print('=== Basic Statistics ===')
print('Mean:', np.mean(scores))
print('Median:', np.median(scores))
print('Standard deviation:', np.std(scores))
print('Variance:', np.var(scores))

# Find:
print()
print('=== Extremes ===')
print('Highest score:', np.max(scores))
print('Lowest score:', np.min(scores))
print('Index of highest:', np.argmax(scores))
print('Index of lowest:', np.argmin(scores))

# Calculate percentiles:
print()
print('=== Percentiles ===')
print('25th percentile (Q1):', np.percentile(scores, 25))
print('50th percentile (Median):', np.percentile(scores, 50))
print('75th percentile (Q3):', np.percentile(scores, 75))

# Generate cumulative:
print()
print('=== Cumulative Operations ===')
print('Cumulative sum of scores:', np.cumsum(scores))

nums = np.array([2, 3, 4, 5])
print('Cumulative product of [2, 3, 4, 5]:', np.cumprod(nums))

# Check:
print()
print('=== Pass/Fail Checks ===')
print('All students passing (score >= 50)?', np.all(scores >= 50))
print('Any student failing (score < 50)?', np.any(scores < 50))
````

````python
# Add missing data
scores_missing = np.array([
    45, 55, 67, np.nan, 88, 91
])

# Calculate:
print('=== Missing Data Demonstration ===')
print('Scores with missing value:', scores_missing)
print()

# Normal mean (what happens?)
normal_mean = np.mean(scores_missing)
print('Normal mean (with NaN):', normal_mean)
print('  -> NaN contaminates the result!')
print()

# nanmean (ignores NaN)
clean_mean = np.nanmean(scores_missing)
print('Nanmean (ignores NaN):', clean_mean)
print('  -> Only valid values are used: (45 + 55 + 67 + 88 + 91) / 5 =', clean_mean)
print()

# Compare the difference
print('=== Comparison ===')
print('With NaN: mean =', np.mean(scores_missing))
print('Without NaN: nanmean =', np.nanmean(scores_missing))
print('Difference: nanmean works, normal mean breaks!')
````
