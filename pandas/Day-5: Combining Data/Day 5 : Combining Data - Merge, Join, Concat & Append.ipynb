# PART 1 — `concat()`

## What is concat?

Stack DataFrames together.

**Signature**

```python
pd.concat(
    objs,
    axis=0,
    ignore_index=False
)
```

## Vertical Concatenation

````python
# Write your Python code here
semester1 = pd.DataFrame({
    "Name" : ["Alice","Bob"],
    "Math" : [90,80]
})

semester2 = pd.DataFrame({
    "Name" : ["Charlie","David"],
    "Math" : [85,75]
})

combined = pd.concat([
    semester1,
    semester2
])

print(combined)
````

**Output**

```
      Name  Math
0    Alice    90
1      Bob    80
0  Charlie    85
1    David    75
```

Notice the index: `0, 1, 0, 1`

## Reset Index

````python
# Write your Python code here
combined = pd.concat([
    semester1,
    semester2,
], ignore_index=True
)

print(combined)
````

**Output**

```
      Name  Math
0    Alice    90
1      Bob    80
2  Charlie    85
3    David    75
```

## Horizontal Concatenation

````python
# Write your Python code here
left_side = pd.DataFrame({
    "A": [1,2]
})

right_side = pd.DataFrame({
    "B": [3,4]
})

print(pd.concat([left_side,right_side], axis=1))
````

**Output**

```
   A  B
0  1  3
1  2  4
```

## `concat()` Summary

| axis | Direction |
|------|-----------|
| `axis=0` | Stack **rows** (↓) |
| `axis=1` | Stack **columns** (→) |

# PART 2 — `merge()`

This is **THE most important function**.

**Signature**

```python
pd.merge(
    left,
    right,
    on=None,
    how="inner"
)
```

## Example

````python
# Write your Python code here
merged_students = pd.merge(students,scores,on="StudentID")
print(merged_students)
````

**Output**

```
   StudentID     Name  Math
0          1    Alice    90
1          2      Bob    85
2          3  Charlie    95
3          4    David    70
```

````python
# Write your Python code here
inner_merge = pd.merge(students,scores,on="StudentID",how="inner")
print(inner_merge)
````
