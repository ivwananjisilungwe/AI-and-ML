## 1. Importing Pandas & Loading Our Dataset

## 2. Selecting One Column

## 3. Selecting Rows — .loc[] & .iloc[]

Pandas has two main row selection systems:
- `.loc[]` — **label** based (uses index labels)
- `.iloc[]` — **position** based (uses integer positions)

| Feature | .loc | .iloc |
|---------|------|-------|
| Based on | Labels | Position |
| Uses | Names | Numbers |
| Example | `df.loc[0, "Age"]` | `df.iloc[0, 1]` |

### 3.1 .loc[] — Label Based Selection

**Signature:** `DataFrame.loc[row_label, column_label]`

````python
# Select row by label
print(df.loc[0,])
````

````python
# Select specific rows
print(df.loc[[0, 2, 4]])
````

````python
# Select row + specific column
print(df.loc[0, "Python"])
````

````python
# Multiple rows and columns
print(df.loc[[0, 2], ["Name", "AI"]])
````

### 3.2 .iloc[] — Position Based Selection

**Signature:** `DataFrame.iloc[row_position, column_position]`

Remember: Python indexing starts at 0.

````python
# First row
print("First row (iloc[0]):")
print(df.iloc[0])
````

````python
# First 3 rows (slicing)
print("First 3 rows:")
print(df.iloc[0:3])
````

````python
# Specific position: first row, second column
print("Row 0, Col 1 (Age of Alice):")
print(df.iloc[0, 1])
````

````python
# Specific rows and columns by position
print("Rows 1-3, Cols 0-2:")
print(df.iloc[1:4, 0:3])
````

---
## 4. Boolean Filtering

This is one of the most important Pandas skills.

**Think:** "Give me students older than 20" → creates a mask of True/False

### 4.1 Simple Condition

````python
# Create a Boolean mask
mask = df["Age"] > 20
print("Boolean mask:")
print(mask)
````

````python
# Use the mask to filter
print("Students older than 20:")
print(df[mask])
````

````python
# Direct (one-liner)
print("Students with Python > 90:")
print(df[df["Python"] > 90])
````

### 4.2 Multiple Conditions (AND)

Use `&` for AND.

**Important:** Always wrap each condition in brackets.
- Correct: `(df["Age"] > 19) & (df["Python"] > 85)`
- Wrong: `df["Age"] > 19 & df["Python"] > 85`

````python
# AND condition: Age > 19 AND Python > 85
filtered = df[(df["Age"] > 19) & (df["Python"] > 85)]
print("Age > 19 AND Python > 85:")
print(filtered)
````

| Operator | Meaning |
|----------|---------|
| `&` | AND |
| `|` | OR |
| `~` | NOT |

### 4.3 OR Conditions

Use `|` for OR.

````python
# OR condition: Math > 90 OR Python > 90
filtered = df[(df["Math"] > 90) | (df["Python"] > 90)]
print("Math > 90 OR Python > 90:")
print(filtered)
````

### 4.4 NOT Condition

Use `~` for NOT (invert a condition).

````python
# NOT: Students NOT older than 20
print("Age <= 20 (NOT older than 20):")
print(df[~(df["Age"] > 20)])
````

---
## 5. Using isin()

Useful when checking if values match **multiple possibilities**.

````python
# Find Alice and Eve
print("Students named Alice or Eve:")
print(df[df["Name"].isin(["Alice", "Eve"])])
````

---
## 6. Sorting Data — sort_values()

**Signature:** `DataFrame.sort_values(by, ascending=True)`

````python
# Sort by Math (ascending)
print("Sorted by Math (low to high):")
print(df.sort_values(by="Math"))
````

````python
# Sort by Math (descending)
print("Sorted by Math (high to low):")
print(df.sort_values(by="Math", ascending=False))
````

````python
# Sort by multiple columns: Age then Math
print("Sorted by Age then Math:")
print(df.sort_values(by=["Age", "Math"]))
````

---
## 7. Ranking — rank()

**Signature:** `Series.rank()`

Assigns ranks (1 = highest, by default).

````python
# Add a ranking column for Python scores
df["Python_Rank"] = df["Python"].rank(ascending=False)
print(df[["Name", "Python", "Python_Rank"]])
````

---
## 8. The query() Method

An alternative way to filter — cleaner for complex conditions.

````python
# Using query instead of brackets
print("Age > 20 (using query):")
print(df.query("Age > 20"))
````

````python
# Multiple conditions in query
print("Age > 19 and Python > 85 (using query):")
print(df.query("Age > 19 and Python > 85"))
````

---
## 9. Creating ML-Style Features

A common ML workflow: create new features, then filter.

````python
# Create an Average score column
df["Average"] = df[["Math", "Python", "AI"]].mean(axis=1)
print(df)
````

````python
# Filter: students with average >= 90
print("Students with Average >= 90:")
print(df[df["Average"] >= 90])
````

## Day 2 Knowledge Check

Answer these questions in your own words:

**1.** What is the difference between `df["Python"]` and `df[["Python"]]`?

**2.** What is the difference between `.loc[]` and `.iloc[]`?

**3.** What does `df["Age"] > 20` return?

**4.** Why must you use brackets around each condition in `df[(cond1) & (cond2)]`?

**5.** How is `df.query()` different from bracket filtering?

**6.** What does `df.sort_values(by="Math", ascending=False)` do?

**1.** `df["Python"]` (single brackets) returns a **Series** — a single column as a 1D array. `df[["Python"]]` (double brackets) returns a **DataFrame** — a 2D table with one column. The difference matters when you need to preserve the 2D structure for operations like column-wise concatenation.

**2.** `.loc[]` is **label-based** — you select rows/columns by their index labels (e.g., `df.loc[0, "Age"]`). `.iloc[]` is **position-based** — you select by integer position like a list (e.g., `df.iloc[0, 1]`). `.loc` includes the endpoint in slices; `.iloc` excludes it (Python-style).

**3.** `df["Age"] > 20` returns a **boolean Series** (True/False) of the same length as the DataFrame. Each element is True if that row's Age > 20, False otherwise. This is called a "Boolean mask".

**4.** You must wrap each condition in brackets because of **Python operator precedence**. The `&` operator binds tighter than comparison operators like `>`. Without brackets, `df["Age"] > 19 & df["Python"] > 85` would be evaluated as `df["Age"] > (19 & df["Python"]) > 85`, which is nonsense (bitwise AND between a number and a Series).

**5.** `df.query()` uses a **string expression** instead of bracket syntax, which is cleaner for complex filters. For example `df.query("Age > 20 & Python > 85")` vs `df[(df["Age"] > 20) & (df["Python"] > 85)]`. You can also reference variables with `@` in the string (e.g., `df.query("Age > @threshold")`). It is slightly slower on small data but scales better and is more readable.

**6.** `df.sort_values(by="Math", ascending=False)` sorts the entire DataFrame by the "Math" column in **descending order** (highest to lowest). By default, `ascending=True` sorts low to high. It returns a new DataFrame by default (does not modify the original unless you set `inplace=True`).

## Day 2 Code Challenge

Dataset:

````python
employees = {
    "Name": ["John", "Mary", "Peter", "Jane", "Mike"],
    "Age": [25, 30, 22, 28, 35],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Salary": [5000, 4000, 4500, 6000, 7000]
}

df = pd.DataFrame(employees)
print(df)
````

**Tasks:**
1. Select only Name and Salary columns
2. Find employees older than 25
3. Find IT department employees
4. Find employees with Age > 25 AND Salary > 5000
5. Sort by Salary highest first
6. Add a Salary rank column

````python
# Your code here:

# 1. Select only Name and Salary
print("Names and Salary")
names_salary = df[["Name", "Salary"]]
print(names_salary)

# 2. Find employees older than 25
print("Employees older than 25")
mask = df["Age"] > 25
print(df[mask])

# 3. Find IT employees
print("It employees:")
it_employees = df[df["Department"].isin(["IT"])]
print(it_employees)

# 4. Age > 25 AND Salary > 5000
print("Age > 25 AND Salary > 5000")
age_and_salary = df[(df["Age"] > 25) & (df["Salary"] > 5000)]
print(age_and_salary)

# 5. Sort by Salary descending
print("Sort by Salary descending")
sort_by_salary = df.sort_values(by="Salary", ascending=False)
print(sort_by_salary)

# 6. Add Salary rank column
df["Salary_Rank"] = df["Salary"].rank(ascending=False)
print(df[["Name", "Salary", "Salary_Rank"]])
````

## Mini Project: Employee Performance Analyzer

Use what you've learned to analyze employee data.

````python
employees = {
    "Name": ["John", "Mary", "Peter", "Jane", "Mike", "Sara", "Tom"],
    "Age": [25, 30, 22, 28, 35, 26, 40],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance"],
    "Salary": [5000, 4000, 4500, 6000, 7000, 4800, 8000],
    "Experience_Years": [2, 5, 1, 4, 8, 3, 12]
}

df = pd.DataFrame(employees)
print(df)
````

**Your program should:**
1. Display IT department employees only
2. Display employees with Salary > 5000
3. Display employees with Experience_Years >= 3 AND Salary > 4500
4. Sort employees by Salary descending and show top 3
5. Add a Salary_Rank column
6. Add a column called "Senior" that is True if Experience_Years >= 5
7. Save the result to `employees_report.csv`
8. Load the CSV again and display

````python
# Your mini project code here:

# 1. IT department employees
print("IT department employees")
employees_in_it = df[df["Department"].isin(["IT"])]
print(employees_in_it)
# 2. Salary > 5000
print("\nSalary > 5000")
mask = df[(df["Salary"] > 5000)]
print(mask)

# 3. Experience >= 3 AND Salary > 4500
print("\nExperience >= 3 AND Salary > 4500")
experience_and_salary = df[(df["Experience_Years"] >= 3) & (df["Salary"] > 4500)]
print(experience_and_salary)

# 4. Sort by Salary descending, top 3
print("\nTop 3 by Salary:")
salary_descending = df.sort_values(by="Salary", ascending=False).head(3)
print(salary_descending)

# 5. Add Salary_Rank
df["Salary_Rank"] = df["Salary"].rank(ascending=False)
print("\nSalary Rank added:")
print(df[["Name", "Salary", "Salary_Rank"]])

# 6. Add Senior column (Experience >= 5)
df["Senior"] = df["Experience_Years"] >= 5
print("\nSenior column (Experience >= 5):")
print(df[["Name", "Experience_Years", "Senior"]])

# 7. Save to CSV
df.to_csv("employees_report.csv", index=False)
print("\nSaved to employees_report.csv")

# 8. Load and display
df_loaded = pd.read_csv("employees_report.csv")
print("\nLoaded from CSV:")
print(df_loaded)
````
