## 2. What is a Pandas Series?

A **Series** is a **one-dimensional labeled array**.

**NumPy:**
```python
arr = np.array([10, 20, 30])
print(arr)  # [10 20 30]
```

**Pandas:**
```python
series = pd.Series([10, 20, 30])
print(series)
```

Output:
```
0    10
1    20
2    30
dtype: int64
```

Pandas gives every value a **label** (an index).

### 2.1 Creating Series - From a List

**Signature:**
```python
pd.Series(data=None, index=None, dtype=None, name=None)
```

````python
# Series from a list
scores = pd.Series([80, 90, 70])
print(scores)
````

### 2.2 Series from NumPy Array

Since Pandas uses NumPy internally:

````python
arr = np.array([100, 200, 300])
series = pd.Series(arr)
print(series)
````

### 2.3 Series with Custom Index

````python
scores = pd.Series(
    [90, 85, 70],
    index=["John", "Mary", "Alice"]
)
print(scores)
````

````python
# Access by label
print(scores["John"])
````

### 2.4 Series from Dictionary

Dictionary keys become the index labels.

````python
student = {
    "Math": 90,
    "Python": 95,
    "Database": 88
}

grades = pd.Series(student)
print(grades)
````

### 2.5 Scalar Series

A single value broadcast across multiple labels.

````python
s = pd.Series(5, index=["a", "b", "c"])
print(s)
````

### 2.6 Series Name

You can give a Series a descriptive name.

````python
scores = pd.Series(
    [90, 80, 70],
    name="Student Scores"
)
print(scores)
````

### 2.7 Series Attributes

````python
s = pd.Series([90, 80, 70], name="Scores")

print("values:", s.values)
print("index:", s.index)
print("dtype:", s.dtype)
print("shape:", s.shape)
print("name:", s.name)
````

## 3. Pandas DataFrame

A **DataFrame** is a 2D table with rows and columns.

**Signature:**
```python
pd.DataFrame(data=None, index=None, columns=None, dtype=None)
```

### 3.1 Creating DataFrames - From Dictionary of Lists

Most common way.

````python
data = {
    "Name": ["John", "Mary", "Alex"],
    "Age": [20, 21, 19],
    "Score": [90, 85, 95]
}

df = pd.DataFrame(data)
print(df)
````

### 3.2 From List of Dictionaries

Very common when working with APIs (JSON responses).

````python
# students = [
#     {"name": "John", "score": 90},
#     {"name": "Mary", "score": 85}
# ]

df = pd.DataFrame(students)
print(df)
````

### 3.3 From NumPy Array

````python
arr = np.array([[1, 2], [3, 4]])
df = pd.DataFrame(arr, columns=["A", "B"])
print(df)
````

### 3.4 Specifying a Custom Index

````python
data = {
    "Name": ["John", "Mary", "Alex"],
    "Score": [90, 85, 95]
}

df = pd.DataFrame(
    data,
    index=["Student1", "Student2", "Student3"]
)
print(df) 
print(pd.Series((2,3,4)))
````

## 4. Pandas Data Types

Pandas automatically detects types:

````python
data = {
    "Name": ["John", "Mary", "Alex"],
    "Age": [20, 21, 19],
    "Score": [90.5, 85.0, 95.5],
    "Passed": [True, True, True]
}
df = pd.DataFrame(data)

print(df.dtypes)
````

## 5. Inspecting DataFrames

This is your daily workflow as a data scientist.

````python
# Create a bigger dataset to inspect
np.random.seed(42)
data = {
    "Name": [f"Student_{i}" for i in range(100)],
    "Age": np.random.randint(18, 30, 100),
    "Score": np.random.uniform(50, 100, 100).round(1),
    "Passed": np.random.choice([True, False], 100)
}
df = pd.DataFrame(data)
print(df)
````

````python
# head() - first n rows (default 5)
# df.head()
````

````python
# tail() - last n rows
df.tail(3)
````

````python
# info() - columns, types, missing values, memory
df.info()
````

````python
# describe() - summary statistics
df.describe()
````

````python
print("shape:", df.shape)
print("columns:", df.columns.tolist())
print("index:", df.index)
print("dtypes:")
print(df.dtypes)
````

````python
# values - returns underlying NumPy array
print(type(df.values))
print(df.values[:3])
````

````python
# sample() - random rows (great for large datasets)
df.sample(5, random_state=42)
````

## 6. Reading & Writing Data Files

Real datasets usually come as files.

### CSV

**Read:**
```python
pd.read_csv("filename.csv")
```
**Save:**
```python
df.to_csv("output.csv", index=False)
```

````python
# Save to CSV
df.to_csv("sample_data.csv", index=False)
print("Saved sample_data.csv")

# Reload it
df_loaded = pd.read_csv("sample_data.csv")
print(df_loaded.head(2))
````

### Other Formats

**Excel:**
```python
pd.read_excel("data.xlsx")
df.to_excel("output.xlsx", index=False)
```

**JSON:**
```python
pd.read_json("data.json")
df.to_json("output.json")
```

**Parquet** (big data):
```python
pd.read_parquet("data.parquet")
df.to_parquet("output.parquet")
```

````python
# memory_usage() - important for large ML datasets
print(df.memory_usage())
````

## Mini Project: Student Performance Analyzer

Create the dataset:

````python
students_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Math": [80, 70, 90, 60, 95],
    "Python": [90, 85, 95, 70, 100],
    "AI": [88, 75, 92, 65, 98]
}
````

**Your program should:**
1. Create a DataFrame
2. Show dataset information (`info()`)
3. Show statistics (`describe()`)
4. Find average scores per subject
5. Save the report to CSV
6. Load the CSV again

````python
# Your mini project code here:
# 1. Create a DataFrame
student_dataframe = pd.DataFrame(students_data)
print("student dataframe:")
print(student_dataframe)

# 2. Show dataset information (`info()`)
print("student info:")
print(student_dataframe.info())

# 3. Show statistics (`describe()`)
print("Statistics:")
print(student_dataframe.describe())

# 4. Find average scores per subject
print("Average:")
student_dataframe["Average"] = student_dataframe[["Math","Python","AI"]].mean(axis=1)
print(student_dataframe)


# 5. Save the report to CSV
student_dataframe.to_csv("students.csv", index=False)
print("saved successifully!!!")

# 6. Load the CSV again
load_csv = pd.read_csv("students.csv")
print(load_csv.head())
````
